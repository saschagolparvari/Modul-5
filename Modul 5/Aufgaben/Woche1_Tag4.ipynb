```python
# Woche 1 – Tag 4: Datenprofiling

# 1. Vollständigkeit
print("1. Fehlende Werte:")
print(df.isna().sum())

# 2. Eindeutigkeit
print("\n2. Doppelte Bestell-IDs:")
print(df["bestell_id"].duplicated().sum())

# 3. Gültigkeit
print("\n3. Ungültige Mengen und Preise:")
print("Menge <= 0:", (df["menge"] <= 0).sum())
print("Preis <= 0:", (df["preis_eur"] <= 0).sum())

# 4. Konsistenz
print("\n4. Kategorien:")
print(df["kategorie"].value_counts(dropna=False))

print("\nLänder:")
print(df["land"].value_counts(dropna=False))

# 5. Plausibilität
print("\n5. Datumsbereich:")
datum_geprueft = pd.to_datetime(df["datum"], errors="coerce", format="mixed")

print("Frühestes Datum:", datum_geprueft.min())
print("Spätestes Datum:", datum_geprueft.max())
print("Nicht lesbare Datumswerte:", datum_geprueft.isna().sum())
```

```python
# Konkrete Beispiele für Datenqualitätsprobleme

print("1. Bestellungen ohne E-Mail:")
display(df[df["kunde_email"].isna()].head(5))

print("2. Bestellungen ohne Kategorie:")
display(df[df["kategorie"].isna()].head(5))

print("3. Doppelte Bestell-IDs:")
display(df[df["bestell_id"].duplicated(keep=False)].sort_values("bestell_id"))

print("4. Ungültige Mengen:")
display(df[df["menge"] <= 0])

print("5. Ungültige Preise:")
display(df[df["preis_eur"] <= 0])

print("6. Uneinheitliche Kategorien:")
display(df[df["kategorie"].isin(["zubehör", "Zubehoer"])])

print("7. Uneinheitliche Länder:")
display(df[df["land"].isin(["Deutschland", "de", "Österreich"])])
```

```python
# Schritt 5: Aggregation und unabhängige Kontrollzählung

# Methode 1: Aggregation mit pandas
ki_ergebnis = df["land"].value_counts()

print("Methode 1 – Aggregation mit pandas:")
print(ki_ergebnis)

# Methode 2: Unabhängige Kontrollzählung
kontrolle = {}

for land in df["land"]:
    if land not in kontrolle:
        kontrolle[land] = 0
    kontrolle[land] += 1

print("\nMethode 2 – Manuelle Kontrollzählung mit Python:")
print(kontrolle)

# Ergebnisse vergleichen
assert ki_ergebnis.to_dict() == kontrolle

print("\nPrüfung erfolgreich: Beide Methoden liefern dieselben Ergebnisse!")
print("Kontrollierte Datensätze:", sum(kontrolle.values()))
```

## Woche 1 – Tag 4: Ergebnisse und Reflexion

### Fünf konkrete Datenqualitätsprobleme

**1. Fehlende Kunden-E-Mail-Adressen**
- Was: Fehlende Werte
- Wo: Spalte `kunde_email`
- Wie viel: 12 von 168 Datensätzen
- Problem: Kunden können nicht zuverlässig per E-Mail kontaktiert werden.

**2. Fehlende Kategorien**
- Was: Unvollständige Produktzuordnung
- Wo: Spalte `kategorie`
- Wie viel: 5 Datensätze
- Problem: Auswertungen nach Produktkategorie können unvollständig sein.

**3. Doppelte Bestellnummern**
- Was: Mehrfach vorkommende Bestell-IDs
- Wo: Spalte `bestell_id`
- Wie viel: 8 zusätzliche Vorkommen
- Problem: Bei einer Auswertung auf Bestellungsebene besteht die Gefahr einer Doppelzählung. Es muss geprüft werden, ob es echte Duplikate oder mehrere Bestellpositionen sind.

**4. Ungültige Mengen und Preise**
- Was: Werte kleiner oder gleich null
- Wo: Spalten `menge` und `preis_eur`
- Wie viel: 3 ungültige Mengen und 3 ungültige Preise
- Problem: Diese Werte können Umsatzberechnungen verfälschen.

**5. Uneinheitliche Länderbezeichnungen**
- Was: Unterschiedliche Schreibweisen für dasselbe Land
- Wo: Spalte `land`
- Wie viel: 4-mal `Deutschland`, 3-mal `de` und 2-mal `Österreich`
- Problem: Auswertungen nach Ländern liefern ohne Vereinheitlichung mehrere Gruppen für dasselbe Land.

### Überprüfung einer KI-gestützten Aggregation

**KI-Vorschlag:** Die Anzahl der Datensätze pro Land mit `value_counts()` berechnen.

**Prüfung:** Eine zweite Zählung wurde unabhängig davon mit einer Python-Schleife durchgeführt.

**Ergebnis:** Beide Methoden lieferten identische Ergebnisse. Insgesamt wurden 168 Datensätze gezählt. Die Übereinstimmung wurde mit `assert` automatisch geprüft.
