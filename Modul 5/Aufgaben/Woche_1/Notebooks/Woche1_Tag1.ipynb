## Datenprofil – Woche 1, Tag 1

### Untersuchte Datei
kickoff_bestellungen.csv

Der Datensatz enthält 20 Zeilen und 7 Spalten.

### Festgestellte Auffälligkeiten

1. **Fehlende Kunden-E-Mails:** Bei 3 von 20 Bestellungen fehlt die E-Mail-Adresse. Dies kann die Kundenkommunikation erschweren.

2. **Ungültige Bestellmengen:** Eine Bestellung enthält die Menge -1 (Bestell-ID 1004). Eine weitere Bestellung hat die Menge 0. Es muss geprüft werden, ob es sich um Fehler oder beispielsweise Retouren handelt.

3. **Uneinheitliche Länderbezeichnungen:** Deutschland wird als DE, de und Deutschland angegeben. Österreich erscheint als AT und Österreich. Dies kann Auswertungen nach Ländern verfälschen.

4. **Problematische Preisformate:** Die Spalte preis_eur wurde als Text eingelesen. Unterschiedliche Dezimaltrennzeichen müssen vor Berechnungen vereinheitlicht werden.
