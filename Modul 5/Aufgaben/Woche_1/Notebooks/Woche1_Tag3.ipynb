```python
df = pd.read_csv(
    "bestellungen_maerz.csv",
    sep=";",
    decimal=",",
    dtype={"plz": "string"}
)

print("Zeilen und Spalten:", df.shape)

print("\nDatentypen:")
print(df.dtypes)

df.head(10)
```

```python
print("Anzahl der Stellen je Postleitzahl:")
print(df["plz"].str.len().value_counts())

print("\nPostleitzahlen mit führender Null:")
print(df.loc[df["plz"].str.startswith("0"), ["ort", "plz"]])

assert df["plz"].str.fullmatch(r"\d{5}").fillna(False).all()

print("\nPrüfung erfolgreich: Alle Postleitzahlen sind fünfstellig!")
```

```python
# Umsatz pro Bestellposition berechnen
df["umsatz"] = df["menge"] * df["preis_eur"]

# Umsatz pro Stadt zusammenfassen
umsatz_pro_stadt = (
    df.groupby("ort")["umsatz"]
    .sum()
    .sort_values(ascending=False)
)

print("Umsatz pro Stadt:")
print(umsatz_pro_stadt.round(2))

print("\nGesamtumsatz:")
print(f"{df['umsatz'].sum():.2f} Euro")
```

```python
# Erwarteter Gesamtumsatz laut Aufgabenstellung
erwarteter_umsatz = 9292.40

# Tatsächlich berechneter Umsatz
berechneter_umsatz = df["umsatz"].sum()

# Automatische Prüfung
assert abs(berechneter_umsatz - erwarteter_umsatz) < 0.01

# Zweite Prüfung: Summe der Städte
assert abs(umsatz_pro_stadt.sum() - erwarteter_umsatz) < 0.01

print("Beide Prüfungen erfolgreich!")
print(f"Gesamtumsatz: {berechneter_umsatz:.2f} €")
```
